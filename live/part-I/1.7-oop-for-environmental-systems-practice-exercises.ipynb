## Exercise 1: A class with state and behaviour

Define a `Glacier` class with `name` and `area_km2` instance attributes, a `describe()` method returning a short string, and a `__repr__`. Create one and print both the object and its description.

````python
# Your solution here

class Glacier:
    def __init__(self, name: str, area_km2: float):
        self.name = name
        self.area_km2 = area_km2

    def describe(self) -> str:
        return f"{self.name} covers {self.area_km2} km²"

    def __repr__(self) -> str:
        return f"Glacier({self.name!r}, {self.area_km2} km²)"

aletsch = Glacier("Aletsch", 78.0)
print(aletsch)
print(aletsch.describe())
````

## Exercise 4: Read and mutate methods, with validation

Define a `Measurements` class holding a list of floats, with `add(x)` to append and `mean()` to return the average. `mean()` should raise `ValueError` if there are no values. Demonstrate both.

````python
# Your solution here

class Measurements:
    def __init__(self):
        self.values: list[float] = []

    def add(self, x: float) -> None:
        self.values.append(x)

    def mean(self) -> float:
        if not self.values:
            raise ValueError("no values")
        return sum(self.values) / len(self.values)

m = Measurements()
try:
    m.mean()
except ValueError as err:
    print("caught:", err)

for x in [1.0, 2.0, 6.0]:
    m.add(x)
print(m.mean())
````

## Exercise 5: Composition

Define a `Catchment` class that holds several `Measurements` objects keyed by name (composition). Add an `add_series(name, series)` method and a `names()` method returning the keys. Build one with two series.

````python
# Your solution here

class Catchment:
    def __init__(self):
        self.series: dict[str, Measurements] = {}

    def add_series(self, name: str, series: Measurements) -> None:
        self.series[name] = series

    def names(self) -> list[str]:
        return list(self.series.keys())

temp = Measurements()
for x in [10.0, 12.0, 14.0]:
    temp.add(x)

precip = Measurements()
for x in [0.0, 5.0, 2.5]:
    precip.add(x)

catchment = Catchment()
catchment.add_series("temperature_celsius", temp)
catchment.add_series("precipitation_mm", precip)
print(catchment.names())
print(catchment.series["temperature_celsius"].mean())
````

## Exercise 7: Inheritance and overriding

Define a base `Station` with a `name` attribute and a `kind()` method returning `"station"`. Define `RiverGauge(Station)` that overrides `kind()` to return `"river gauge"`. Show that a `RiverGauge` keeps the inherited name but reports the new kind.

````python
# Your solution here

class Station:
    def __init__(self, name: str):
        self.name = name

    def kind(self) -> str:
        return "station"

class RiverGauge(Station):
    def kind(self) -> str:
        return "river gauge"

s = Station("Generic")
g = RiverGauge("Rhone at Porte du Scex")
print(s.name, "->", s.kind())
print(g.name, "->", g.kind())
````

## Exercise 8: An invariant with assert

Write `normalise(weights)` that divides each weight by the total, and asserts the invariant that the result sums to 1 (within a small tolerance). Test it on `[2.0, 2.0]`.

````python
# Your solution here

def normalise(weights: list[float]) -> list[float]:
    total = sum(weights)
    result = []
    for w in weights:
        result.append(w / total)
    assert abs(sum(result) - 1.0) < 1e-9, "normalisation failed"
    return result

print(normalise([2.0, 2.0]))
````

## Exercise 9: A custom exception

Define `NegativeDischargeError(ValueError)` and a function `check(q_m3s)` that raises it when discharge is negative. Catch it and print the message.

````python
# Your solution here

class NegativeDischargeError(ValueError):
    pass

def check(q_m3s: float) -> None:
    if q_m3s < 0:
        raise NegativeDischargeError(f"negative discharge: {q_m3s} m3/s")

try:
    check(-5.0)
except NegativeDischargeError as err:
    print("caught:", err)
````

## Exercise 10: Try / except / else / finally

Write `safe_divide(a, b)` that returns `a / b`, catches `ZeroDivisionError` (the built-in exception Python raises when dividing by zero), returns `None` in that case, prints a message in the `else` branch when it succeeds, and always prints "done" in `finally`. Call it with `(6, 2)` and `(6, 0)`.

````python
# Your solution here

def safe_divide(a: float, b: float) -> float | None:
    try:
        result = a / b
    except ZeroDivisionError:
        return None
    else:
        print("division succeeded")
        return result
    finally:
        print("done")

print(safe_divide(6, 2))
print(safe_divide(6, 0))
````

## Exercise 11: Validate preconditions

Write `validate(temp_celsius, rh_percent)` that raises `ValueError` if the temperature is below absolute zero or the relative humidity is outside 0–100 %. Show it passing on valid input and raising on `rh_percent = 150`.

````python
# Your solution here

def validate(temp_celsius: float, rh_percent: float) -> None:
    if temp_celsius < -273.15:
        raise ValueError("temperature below absolute zero")
    if rh_percent < 0.0 or rh_percent > 100.0:
        raise ValueError("relative humidity must be between 0 and 100 %")

validate(18.0, 60.0)          # valid input passes silently
try:
    validate(18.0, 150.0)
except ValueError as err:
    print("caught:", err)
````

## Exercise 13: Write and run a pytest suite

Write a module `mymod.py` with `double(x)` returning `2 * x`, and a parametrised test file that checks three cases. Run pytest on it with a subprocess and print the output.

````python
# Your solution here

from pathlib import Path
import subprocess
import sys

Path("_files").mkdir(exist_ok=True)

Path("_files/mymod.py").write_text("""
def double(x):
    return 2 * x
""", encoding="utf-8")

Path("_files/test_mymod.py").write_text("""
import pytest
from mymod import double

@pytest.mark.parametrize("x, expected", [(0, 0), (2, 4), (-3, -6)])
def test_double(x, expected):
    assert double(x) == expected
""", encoding="utf-8")

result = subprocess.run(
    [sys.executable, "-m", "pytest", "test_mymod.py", "-q", "--color=no"],
    capture_output=True, text=True, cwd="_files",
)
print((result.stdout + result.stderr).strip())
````
